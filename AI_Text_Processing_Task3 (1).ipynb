```python

# TASK 3: PYTHON FOR AI APPLICATIONS

import re
import pandas as pd
from collections import Counter
import os

print("Libraries imported successfully!")
```

```python

# FUNCTION 1: CLEAN TEXT

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    return text


# FUNCTION 2: PROCESS TEXT

def process_text(text):

    cleaned_text = clean_text(text)

    # Store words in a list
    words = cleaned_text.split()

    # Dictionary to store word frequency
    word_frequency = {}

    for word in words:
        if word in word_frequency:
            word_frequency[word] += 1
        else:
            word_frequency[word] = 1

    # Count sentences
    sentences = [
        s.strip()
        for s in re.split(r'[.!?]+', text)
        if s.strip()
    ]

    # Store structured information
    result = {
        "Total Words": len(words),
        "Unique Words": len(set(words)),
        "Total Sentences": len(sentences),
        "Word Frequency": word_frequency
    }

    return result


print("Text processing functions created successfully!")
```

```python

# FUNCTION 3: DISPLAY ANALYSIS

def display_result(result):

    print("\n========== TEXT ANALYSIS ==========")

    print("Total Words:", result["Total Words"])
    print("Unique Words:", result["Unique Words"])
    print("Total Sentences:", result["Total Sentences"])

    print("\n========== WORD FREQUENCY ==========")

    for word, count in result["Word Frequency"].items():
        print(word, ":", count)

    print("\n========== TOP 5 WORDS ==========")

    top_words = Counter(
        result["Word Frequency"]
    ).most_common(5)

    for word, count in top_words:
        print(word, ":", count)


print("Analysis function created successfully!")
```

```python

# FUNCTION 4: STRUCTURED DATA

def create_dataframe(result):

    data = []

    for word, count in result["Word Frequency"].items():

        data.append({
            "Word": word,
            "Frequency": count
        })

    df = pd.DataFrame(data)

    df = df.sort_values(
        by="Frequency",
        ascending=False
    )

    return df


print("Pandas structured output function created!")
```

```python

# FUNCTION 5: SAVE REPORT

def save_report(df, result):

    os.makedirs("output", exist_ok=True)

    # Save word frequency
    df.to_csv(
        "output/word_analysis.csv",
        index=False
    )

    # Save text summary
    summary = pd.DataFrame([
        {"Metric": "Total Words", "Value": result["Total Words"]},
        {"Metric": "Unique Words", "Value": result["Unique Words"]},
        {"Metric": "Total Sentences", "Value": result["Total Sentences"]}
    ])

    summary.to_csv(
        "output/text_summary.csv",
        index=False
    )

    print("Reports saved successfully!")


print("Report function created successfully!")
```

```python
# MAIN APPLICATION - GOOGLE COLAB INTERFACE

import ipywidgets as widgets
from IPython.display import display, clear_output

print("======================================")
print(" AI TEXT PROCESSING APPLICATION")
print("======================================")

# Create text input box
text_box = widgets.Textarea(
    placeholder="Enter your paragraph here...",
    description="Input:",
    layout=widgets.Layout(width="90%", height="130px")
)

# Create button
analyze_button = widgets.Button(
    description="Analyze Text",
    button_style="success"
)

# Output area
output_area = widgets.Output()


# Button click function
def analyze_clicked(button):

    with output_area:
        clear_output()

        user_input = text_box.value

        if not user_input.strip():
            print("Please enter valid text.")

        else:
            result = process_text(user_input)

            display_result(result)

            df = create_dataframe(result)

            print("\n========== STRUCTURED OUTPUT ==========")
            display(df)

            # Generate word frequency chart
            generate_chart(result["Word Frequency"])

            save_report(df, result)

            print("\nAnalysis completed successfully!")


analyze_button.on_click(analyze_clicked)

# Display application
display(text_box, analyze_button, output_area)
```

```python

import matplotlib.pyplot as plt

def generate_chart(frequency):

    top_words = Counter(frequency).most_common(10)

    if not top_words:
        print("No words available for chart.")
        return

    words = [item[0] for item in top_words]
    counts = [item[1] for item in top_words]

    plt.figure(figsize=(10, 5))
    plt.bar(words, counts)

    plt.title("Top 10 Word Frequency Analysis")
    plt.xlabel("Words")
    plt.ylabel("Frequency")

    plt.xticks(rotation=30)
    plt.tight_layout()

    plt.show()
```
